```python
datasets = {}

for year in range(2020, 2025):
    file = f"ttc-bus-delay-data-{year}.xlsx"
    sheets = pd.read_excel(file, sheet_name=None)

    datasets[year] = pd.concat(
        sheets.values(), ignore_index=True
    )

datasets[2025] = pd.read_csv(
    "TTC Bus Delay Data since 2025.csv",
    low_memory=False
)

# Display dataset dimensions
for year, df in datasets.items():
    print(f"{year}: {df.shape[0]} rows, {df.shape[1]} columns")
```

```python
# Create a list to store datasets
all_datasets = []

for year, df in datasets.items():

    # Create a copy to preserve the original data
    temp_df = df.copy()

    # Add year column
    temp_df['Year'] = year

    # Add dataset to the list
    all_datasets.append(temp_df)

# Combine all datasets
df_combined = pd.concat(
    all_datasets,
    ignore_index=True,
    sort=False
)

# Display integration results
print("Data integration completed successfully!")

print("\nTotal rows:", df_combined.shape[0])
print("Total columns:", df_combined.shape[1])

print("\nRecords by year:")
print(df_combined['Year'].value_counts().sort_index())

print("\nCombined columns:")
print(df_combined.columns.tolist())
```

# **6. Data Type & Time Standardization**

```python
print("Dataset Information:")
df_cleaned.info()

print("\nData Types:")
print(df_cleaned.dtypes)

print("\nSample Records:")
display(df_cleaned.head())
```

```python
import re

# Save backup
df_before_formatting = df_cleaned.copy()

# Function to standardize identifiers
def standardize_identifier(value):

    value = str(value).strip()

    # Remove .0 only from numeric identifiers
    if re.fullmatch(r'\d+\.0', value):
        value = value[:-2]

    return value

# Apply standardization
df_cleaned['Route'] = (
    df_cleaned['Route']
    .apply(standardize_identifier)
    .astype('string')
)

df_cleaned['Vehicle'] = (
    df_cleaned['Vehicle']
    .apply(standardize_identifier)
    .astype('string')
)

print("Route and Vehicle Formatting Completed!")

print("\nSample Routes:")
print(df_cleaned['Route'].head(10).tolist())

print("\nSample Vehicles:")
print(df_cleaned['Vehicle'].head(10).tolist())

print("\nDataset Shape:", df_cleaned.shape)
```

```python
# Check types stored in Time column
print("Time Value Types:")
print(df_cleaned['Time'].apply(type).value_counts())

# Convert time values for validation
time_values = pd.to_datetime(
    df_cleaned['Time'].astype(str),
    format='%H:%M:%S',
    errors='coerce'
)

# Count invalid times
invalid_times = time_values.isna().sum()

print("\nInvalid Time Values:", invalid_times)

# Display examples of invalid times
print("\nSample Invalid Times:")
display(
    df_cleaned.loc[
        time_values.isna(),
        ['Year', 'Date', 'Time']
    ].head(15)
)

print("\nTotal Records:", len(df_cleaned))
```

```python
import datetime

# Save backup
df_before_time_cleaning = df_cleaned.copy()

def standardize_time(value):

    # Handle datetime objects
    if isinstance(value, datetime.datetime):
        return value.strftime('%H:%M:%S')

    # Handle time objects
    if isinstance(value, datetime.time):
        return value.strftime('%H:%M:%S')

    # Handle string values
    value = str(value).strip()

    # HH:MM -> HH:MM:SS
    if len(value) == 5 and value.count(':') == 1:
        value += ':00'

    # Validate time
    try:
        return datetime.datetime.strptime(
            value, '%H:%M:%S'
        ).strftime('%H:%M:%S')

    except ValueError:
        return pd.NA

# Apply standardization
df_cleaned['Time'] = (
    df_cleaned['Time']
    .apply(standardize_time)
    .astype('string')
)

print("Time Standardization Completed!")

print("\nMissing/Invalid Times:",
      df_cleaned['Time'].isna().sum())

print("\nSample Times:")
print(df_cleaned['Time'].head(10).tolist())

print("\nDataset Shape:", df_cleaned.shape)
```

# **7. Outlier Detection**

```python
# Descriptive statistics
print("Descriptive Statistics:")
display(df_cleaned[['Min Delay', 'Min Gap']].describe())

# Check negative and zero values
for col in ['Min Delay', 'Min Gap']:

    print(f"\n{col}:")

    print("Negative values:",
          (df_cleaned[col] < 0).sum())

    print("Zero values:",
          (df_cleaned[col] == 0).sum())

    # Calculate IQR
    Q1 = df_cleaned[col].quantile(0.25)
    Q3 = df_cleaned[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Count outliers
    outliers = (
        (df_cleaned[col] < lower_bound) |
        (df_cleaned[col] > upper_bound)
    ).sum()

    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Potential Outliers:", outliers)

# Display largest delays
print("\nTop 10 Largest Delays:")
display(
    df_cleaned.nlargest(10, 'Min Delay')[
        ['Year', 'Date', 'Route', 'Incident',
         'Min Delay', 'Min Gap']
    ]
)
```

```python
# Define extreme delays as more than 180 minutes
extreme_delays = df_cleaned[
    df_cleaned['Min Delay'] > 180
]

# Count extreme delays by year
extreme_by_year = (
    extreme_delays.groupby('Year')
    .size()
    .reindex(sorted(df_cleaned['Year'].unique()), fill_value=0)
)

print("Extreme Delays (>180 minutes) by Year:")
display(extreme_by_year.to_frame('Extreme Delay Count'))

# Show maximum delay per year
print("\nMaximum Delay by Year:")
display(
    df_cleaned.groupby('Year')['Min Delay']
    .max()
    .to_frame('Maximum Delay')
)

# Most common incidents among extreme delays
print("\nTop Incidents for Extreme Delays:")
display(
    extreme_delays['Incident']
    .value_counts()
    .head(10)
)

print("\nTotal Extreme Delays:", len(extreme_delays))
```

```python
# Count specific high delay values
print("High Delay Value Frequencies:")

high_delay_counts = (
    df_cleaned[
        df_cleaned['Min Delay'] >= 900
    ]['Min Delay']
    .value_counts()
    .sort_index()
)

display(high_delay_counts.to_frame('Count'))

# Check 999 values by year
print("\n999-Minute Delays by Year:")

delay_999 = df_cleaned[
    df_cleaned['Min Delay'] == 999
]

display(
    delay_999.groupby('Year')
    .size()
    .to_frame('Count')
)

# Show sample records
print("\nSample 999-Minute Delays:")

display(
    delay_999[
        ['Year', 'Date', 'Route', 'Incident',
         'Incident Code', 'Min Delay', 'Min Gap']
    ].head(15)
)

print("\nTotal 999-Minute Delays:", len(delay_999))
```

```python
# Count 999 in each column
delay_999 = (df_cleaned['Min Delay'] == 999).sum()
gap_999 = (df_cleaned['Min Gap'] == 999).sum()

# Count records where both equal 999
both_999 = (
    (df_cleaned['Min Delay'] == 999) &
    (df_cleaned['Min Gap'] == 999)
).sum()

print("Min Delay = 999:", delay_999)
print("Min Gap = 999:", gap_999)
print("Both = 999:", both_999)

# Check incidents associated with 999
suspicious_records = df_cleaned[
    (df_cleaned['Min Delay'] == 999) |
    (df_cleaned['Min Gap'] == 999)
]

print("\nIncidents associated with 999:")
display(
    suspicious_records.groupby(
        ['Year', 'Incident'],
        dropna=False
    ).size().to_frame('Count')
)

print("\nTotal affected records:", len(suspicious_records))
```

```python
# Save backup
df_before_outlier_handling = df_cleaned.copy()

# Flag records containing 999
df_cleaned['Suspicious_999'] = (
    (df_cleaned['Min Delay'] == 999) |
    (df_cleaned['Min Gap'] == 999)
)

print("Suspicious Value Flagging Completed!")

print("\nFlagged records:",
      df_cleaned['Suspicious_999'].sum())

print("\nUnflagged records:",
      (~df_cleaned['Suspicious_999']).sum())

print("\nDataset Shape:", df_cleaned.shape)

print("\nFlag Distribution:")
display(
    df_cleaned['Suspicious_999']
    .value_counts()
    .to_frame('Count')
)
```
